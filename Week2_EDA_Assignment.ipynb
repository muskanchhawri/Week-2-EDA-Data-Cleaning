```python

import pandas as pd

# =====================================
# PART 1: LOAD CSV DATA
# =====================================

df = pd.read_csv("student_data.csv")

print("First 5 rows:")
print(df.head())

# =====================================
# PART 2: FILTERING AND SORTING
# =====================================

print("\nStudents with Marks above 80:")
print(df[df["Marks"] > 80])

print("\nStudents sorted by Marks:")
print(df.sort_values("Marks", ascending=False)[["Name", "Marks"]])

print("\nStudent with highest Marks:")
print(df.loc[df["Marks"].idxmax(), ["Name", "Marks"]])

# =====================================
# PART 3: BASIC STATISTICS
# =====================================

print("\nAverage Marks:", df["Marks"].mean())
print("Maximum Marks:", df["Marks"].max())
print("Minimum Marks:", df["Marks"].min())

# =====================================
# PART 4: EXPLORATORY DATA ANALYSIS
# =====================================

print("\nDataset Shape:")
print(df.shape)

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
print(df.describe())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

# =====================================
# PART 5: CLEAN CSV DATA
# =====================================

df = df.drop_duplicates()

print("\nData after removing duplicates:")
print(df)

print("\nDuplicate count after cleaning:")
print(df.duplicated().sum())

# =====================================
# PART 6: MISSING VALUES PRACTICE
# =====================================

data = {
    "Name": ["Aman", "Riya", "Karan", "Neha"],
    "Age": [20, 19, None, 21],
    "Marks": [85, None, 67, 88]
}

sample_df = pd.DataFrame(data)

print("\nOriginal Sample Data:")
print(sample_df)

print("\nMissing Values in Sample Data:")
print(sample_df.isnull().sum())

# Fill missing values with column averages
sample_df["Age"] = sample_df["Age"].fillna(
    sample_df["Age"].mean()
)

sample_df["Marks"] = sample_df["Marks"].fillna(
    sample_df["Marks"].mean()
)

print("\nSample Data after Filling Missing Values:")
print(sample_df)

print("\nMissing Values after Filling:")
print(sample_df.isnull().sum())
```
